```python
import os
import pandas as pd

# Carpeta de origen
origen = r"C:\Users\RGARCIA\Downloads\00 Metadatas\Apsa\260506"

def unir_txt(directorio, archivo_salida):
    todos_los_datos = []
    columnas_base = None

    for archivo in os.listdir(directorio):
        if archivo.endswith('.txt'):
            ruta_archivo = os.path.join(directorio, archivo)
            
            # Leer archivo con separador "~"
            df = pd.read_csv(ruta_archivo, delimiter='~', header=0, encoding='utf-8')
            
            # Guardar columnas de referencia en la primera iteración
            if columnas_base is None:
                columnas_base = df.columns
            
            # Asegurar que no se repitan encabezados
            df.columns = columnas_base
            
            # Agregar columna con nombre del archivo
            df['Archivo'] = archivo
            todos_los_datos.append(df)

    # Unir todos los DataFrames
    df_final = pd.concat(todos_los_datos, ignore_index=True)

    # Convertir columnas numéricas si existen
    if 'Estatus' in df_final.columns:
        df_final['Estatus'] = pd.to_numeric(df_final['Estatus'], errors='coerce')
    if 'Monto' in df_final.columns:
        df_final['Monto'] = pd.to_numeric(df_final['Monto'], errors='coerce')

    # Ordenar por la columna 'FechaEmision' si existe
    if 'FechaEmision' in df_final.columns:
        df_final = df_final.sort_values(by='FechaEmision', ascending=True)

    # Exportar a TXT con separador "~"
    df_final.to_csv(archivo_salida, sep='~', index=False, encoding='utf-8')
    print(f"Archivo TXT '{archivo_salida}' generado exitosamente.")

# Ejecutar función
directorio_txt = origen
archivo_txt_salida = fr"{origen}\Metadata.txt"

unir_txt(directorio_txt, archivo_txt_salida)
```
